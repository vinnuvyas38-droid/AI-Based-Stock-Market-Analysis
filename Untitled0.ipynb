```python
!pip install yfinance plotly scikit-learn -q

import yfinance as yf
import pandas as pd
import numpy as np
import plotly.graph_objects as go

ticker = "TCS.NS"                                   # change to AAPL, RELIANCE.NS, etc.
df = yf.download(ticker, period="3y", auto_adjust=True)
df.columns = df.columns.get_level_values(0)         # flatten column headers
df = df.dropna()                                    # remove rows with missing values

print(df.shape)
print(df.tail())
print(df.isnull().sum())
```

```python
# Moving averages (average price over last 20 / 50 days)
df["MA20"] = df["Close"].rolling(20).mean()
df["MA50"] = df["Close"].rolling(50).mean()

# RSI (14-day): measures if a stock is overbought (>70) or oversold (<30)
delta = df["Close"].diff()
gain = delta.clip(lower=0).rolling(14).mean()
loss = (-delta.clip(upper=0)).rolling(14).mean()
df["RSI"] = 100 - (100 / (1 + gain / loss))

# Daily returns and 30-day annualised volatility (risk measure)
df["Return"] = df["Close"].pct_change()
df["Volatility"] = df["Return"].rolling(30).std() * np.sqrt(252) * 100

# Interactive chart
fig = go.Figure()
fig.add_trace(go.Scatter(x=df.index, y=df["Close"], name="Close"))
fig.add_trace(go.Scatter(x=df.index, y=df["MA20"], name="MA20"))
fig.add_trace(go.Scatter(x=df.index, y=df["MA50"], name="MA50"))
fig.update_layout(title=f"{ticker} Price with Moving Averages")
fig.show()

print(df[["Close", "MA20", "MA50", "RSI", "Volatility"]].tail())
```

```python
# Take the most recent row of data
latest = df.iloc[-1]

price = latest["Close"]
ma20 = latest["MA20"]
ma50 = latest["MA50"]
rsi = latest["RSI"]
vol = latest["Volatility"]

insights = []

# 1. Trend rule: compare price and moving averages
if price > ma50 and ma20 > ma50:
    insights.append("TREND: Bullish. Price is above the 50-day average and the short-term trend is rising.")
elif price < ma50 and ma20 < ma50:
    insights.append("TREND: Bearish. Price is below the 50-day average and the short-term trend is falling.")
else:
    insights.append("TREND: Mixed or sideways. Moving averages are giving conflicting signals.")

# 2. RSI rule: overbought / oversold
if rsi > 70:
    insights.append(f"MOMENTUM: RSI is {rsi:.1f}. The stock may be overbought (risen too fast).")
elif rsi < 30:
    insights.append(f"MOMENTUM: RSI is {rsi:.1f}. The stock may be oversold (fallen too much).")
else:
    insights.append(f"MOMENTUM: RSI is {rsi:.1f}. Momentum is in the neutral zone.")

# 3. Risk rule: based on annualised volatility
if vol > 35:
    risk = "HIGH"
elif vol > 20:
    risk = "MEDIUM"
else:
    risk = "LOW"
insights.append(f"RISK: {risk}. Annualised volatility is {vol:.1f}%.")

# 4. Recent performance
change_30d = (price / df["Close"].iloc[-31] - 1) * 100
insights.append(f"PERFORMANCE: The stock moved {change_30d:+.1f}% over the last 30 trading days.")

print(f"=== Insights for {ticker} ===")
for line in insights:
    print("-", line)
print("\nDisclaimer: This is a decision-support tool, not financial advice.")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

# ---- 1. Build features (inputs the model learns from) ----
data = df.copy()
data["Lag1"] = data["Close"].shift(1)    # yesterday's close
data["Lag2"] = data["Close"].shift(2)    # close 2 days ago
data["Lag3"] = data["Close"].shift(3)    # close 3 days ago

# ---- 2. Target: tomorrow's closing price ----
data["Target"] = data["Close"].shift(-1)

# The last row has no tomorrow, and early rows have empty indicators
data = data.dropna()

features = ["Close", "Lag1", "Lag2", "Lag3", "MA20", "MA50", "RSI", "Volatility", "Volume"]
X = data[features]
y = data["Target"]

# ---- 3. Split by TIME: first 80% train, last 20% test ----
split = int(len(data) * 0.8)
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]

# ---- 4. Train both models ----
lr = LinearRegression()
lr.fit(X_train, y_train)

rf = RandomForestRegressor(n_estimators=200, random_state=42)
rf.fit(X_train, y_train)

# ---- 5. Predict on the test set ----
lr_pred = lr.predict(X_test)
rf_pred = rf.predict(X_test)

# ---- 6. Evaluate: MAE and RMSE (lower = better) ----
def evaluate(name, actual, predicted):
    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    print(f"{name}: MAE = {mae:.2f}, RMSE = {rmse:.2f}")

print(f"Training days: {len(X_train)}, Testing days: {len(X_test)}\n")
evaluate("Linear Regression", y_test, lr_pred)
evaluate("Random Forest   ", y_test, rf_pred)
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=y_test.index, y=y_test, name="Actual"))
fig.add_trace(go.Scatter(x=y_test.index, y=lr_pred, name="Linear Regression"))
fig.add_trace(go.Scatter(x=y_test.index, y=rf_pred, name="Random Forest"))
fig.update_layout(title=f"{ticker}: Actual vs Predicted Close (Test Period)")
fig.show()
```

```python
# Forecast the next 5 trading days using Linear Regression (the better model)
# Retrain on ALL available data first
lr_final = LinearRegression()
lr_final.fit(X, y)

# Start from a copy of the close price history
history = df["Close"].copy()
volume_last = df["Volume"].iloc[-1]
forecasts = []

for i in range(5):
    close_series = history

    # Recompute indicators from the updated history
    ma20 = close_series.rolling(20).mean().iloc[-1]
    ma50 = close_series.rolling(50).mean().iloc[-1]

    d = close_series.diff()
    g = d.clip(lower=0).rolling(14).mean().iloc[-1]
    l = (-d.clip(upper=0)).rolling(14).mean().iloc[-1]
    rsi_val = 100 - (100 / (1 + g / l))

    vol_val = close_series.pct_change().rolling(30).std().iloc[-1] * np.sqrt(252) * 100

    row = pd.DataFrame([{
        "Close": close_series.iloc[-1],
        "Lag1": close_series.iloc[-2],
        "Lag2": close_series.iloc[-3],
        "Lag3": close_series.iloc[-4],
        "MA20": ma20, "MA50": ma50,
        "RSI": rsi_val, "Volatility": vol_val,
        "Volume": volume_last
    }])[features]

    next_price = lr_final.predict(row)[0]
    forecasts.append(next_price)

    # Add the predicted price to history so the next day can use it
    next_date = history.index[-1] + pd.tseries.offsets.BDay(1)
    history = pd.concat([history, pd.Series([next_price], index=[next_date])])

forecast_dates = history.index[-5:]
forecast_df = pd.DataFrame({"Date": forecast_dates, "Predicted Close": np.round(forecasts, 2)})
print(forecast_df.to_string(index=False))

# Plot last 90 days + forecast
fig = go.Figure()
fig.add_trace(go.Scatter(x=df.index[-90:], y=df["Close"].iloc[-90:], name="Historical Close"))
fig.add_trace(go.Scatter(x=forecast_dates, y=forecasts, name="5-Day Forecast",
                         mode="lines+markers", line=dict(dash="dash")))
fig.update_layout(title=f"{ticker}: 5-Day Forecast")
fig.show()
```
